```python
import torch
from transformers import AutoModel
```

```python
model_dir = "/software/huggingface/hub/models--ctheodoris--Geneformer/snapshots/04c2b2e84da7c0f385c3f9ad8f3ec24bab6650e5/Geneformer-V2-104M/"

# 1. Carica il modello
model = AutoModel.from_pretrained(model_dir)
```

```python
# --- CONTROLLO 1: Modalità del Modello (eval vs train) ---
print(f"Il modello è in training mode? {model.training}")

# Impostiamo la modalità eval
model.eval()
print(f"Dopo model.eval() -> Il modello è in eval mode? {not model.training}")


# --- CONTROLLO 2: Stato dei Gradienti (torch.no_grad) ---
print(f"\nI gradienti sono abilitati normalmente? {torch.is_grad_enabled()}")

# Verifichiamo dentro il contesto no_grad
with torch.no_grad():
    print(
        f"Dentro 'with torch.no_grad()' -> I gradienti sono disabilitati? {not torch.is_grad_enabled()}"
    )
```
